# Typecasting in python
**The conversion of datatype is called typecasting<br>Like :**
- Integer -> String
- String -> Integer
- Integer -> Float
- Float -> Integer
- Flot -> String etc

```python
# Integer to String
age = 23
print(type(age))
age = str(age)
print(type(age))
```

```python
# String to Integer
age = str(23)
print(type(age))
age = int(age)
print(type(age))
```

```python
# Float to Integer
marks = 87.6
print(marks)
print(type(marks))
marks = int(marks) #When changing from float to integer, the python removes the decimal values so always the floor value is taken
print(type(marks))
print(marks)
```
