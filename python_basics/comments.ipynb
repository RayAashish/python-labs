# Comments in Python
**Comments are totally ignored by Python** <br>
~ In python there are two types of comments :
1. Single Line Comment
-> Anything written infront of # is comment in python
2. Multiline Comment
-> Anything written under """ comment """ is a comment

```python
age = int(input("Enter age : ")) # Single line comment
print(age)
```

```python
"""
This is multi line comment
name is taken as the input
"""
name = input("Enter name : ")
print(name)
```
