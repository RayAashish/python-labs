# Python Input/Output
**In python print() function is used for output & input() function is used for input** 

### Input always return the value in string format so it must should be typecasted to work on with


```python
# Output
var = "Something something"
print(var)
```

```python
# Multiple outputs
print("hello", "world", "from", "python")
```

```python
# End Seprator
print("hello", end = " | ")
print("world")
```

```python
# Input
name = input("Enter name: ")
print(name)
print(type(name))
```

```python
# Input
age = input("Enter age: ")
print(age)
print(type(age))
```

```python
# Type casting of input
age = int(input("Enter age : "))
print(age)
print(type(age))
```
