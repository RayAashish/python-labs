# Python Basics
## 1. Variables
**A variable is a named container which stores a value**
```Python
# Variable Example
name = "Aashish"
age = 23
```
## Important Concepts : 
1. Unlike Java, it doesn't require a explict datatype declaration.
2. Multiple variable can be assigned at a time
3. Multiple variable can hold same value

````python
# Example of 1
name = "Aashish"
age = 23
print(type(name))
print(type(age))
````

````python
# Multiple variable assignment
a, b, c = 10, 23, 45
print('a : ', a)
print('b : ', b)
print('c : ', c)
````

````python
# Example of 3
x = y = z = 23
print('x : ', x)
print('y : ', y)
print('z : ', z)
````

# Dynamic Datatypes
**Datatypes in python changes dynamically.<br> For example if a variable is assigned with int type & later string is assigned to that variable, the variable will change at the runtime<br>Example below :**

````python
x = 23
print(type(x))
x = "Aashish"
print(type(x))
````
