# f-string in Python
**f-strings was introduced in Python 3.6<br>f-string are used for embedding variables inside the string**

```python
name = 'Aashish Kr Ray'
age = 23
```

```python
# Tradational Method
print("My name is " +  name + " & I am "  +str(age) + " years old")
```

```python
#f-string method
print(f"My name is {name} & I am {age} years old")
```

# Expression using f-string

```python
x = 21
y = 23
print(f"Sum : {x + y}")
```

# Function using f-string

```python
surname = "ray"
first_name = "AASHISH"
print(f"First_Name : {first_name.lower()}")
print(f"Surname : {surname.upper()}")
```

```python
pi = 3.141596
print(f"pi : {pi:.2f}")
```

```python
name = "Aashish"
print(f"|{name : < 20}|")
```
