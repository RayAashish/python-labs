**~ Everything in python is an object**
## Types of datatype
### Numeric
1. int : Any +ve|-ve whole number
2. float : Any +ve|-ve decimal number
3. string : Any value enclosed in '', "", """""", '''''' (Simply any text)
4. boolean : True | False
5. complex : Used in mathematics
### Collection Type
1. List : Dynamic array (Alternative of ArrayList of Java)
2. Tuple : Immutable dynamic array
3. Set : Holds the unique number
4. Dictonary : Holds data in key value pairs (Alternative of Hashmap of Java)

```python
# Tuple
coordinates = (20, 90)
print(coordinates)
print(type(coordinates))
# coordinates[0] = 12 # It will throw an error because lists are immutable
```

```python
# Set
set = {12, 19, 12, 12, 12, 19, 27}
print(set)
print(type(set))
```

```python
# Dictonary
dict = {
    "name" = "Aashish",
    age = 23 }
print(dict)
print(dict.values())
print(dict.keys())
print(type(dict))
```
