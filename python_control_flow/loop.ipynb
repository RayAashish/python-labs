# Loop
**Loop is used for repeating the code <br>There are only two types of loop in Python unlike 3 in java**
1. for loop : Primarly used when loop when range (no. of iterations) is known
```python
    # Syntax
    for i in range:
        # loop body
```
3. while loop : Primarly used when range is unknown
```python
    # Syntax
    while (condition):
        # loop body
```

````python
# Example of loop
for i in range (10):
    print(f"{i}", end = ' ')
````

````python
 # Loop through a string
name = "Aashish Kumar Ray"
for ch in name : 
    print(ch, end = '|')
````

````python
# Loop through a list
nums = [1, 2, 0, -3, 8, -9]
nums.sort()
for n in nums :
    print(n, end = ', ')
````

# Important concepts :: Understanding range()
## 1. range(start, end)
**When a single number is pushed as range parameter, 0 acts as the starting & 1 is the step count in such condition. But we can explictely paramterised the range with start & end.**


````python
 # Example
for i in range (5, 10):
    print(i, end = ' | ')
````

## 2. range (start, end, step)

````python
# Example
for i in range (0, 10, 2):
    print(i, end = ", ")
````

## 3. reverse loop
```python
for i in range(10, 0, -1):
    print(i)
```

````python
for i in range(10, 0, -1):
    print(i, end = ', ')
````
