# Match Case
**Math case is the alternative of switch case of java. It was introduced in Python 3.1<br>**
1. Default case is denoted by this :
```python
       case _:
```
2. Multiple cases can also be used using like this :
```python
   case a | b | c | d:
```

````python
day = 2
match day:
    case 1:
        print("Sunday")
    case 2:
        print("Monday")
````

````python
# With default case 
day = 10
match day :
    case 1:
        print("Sunday")
    case 2:
        print("Monday")
    case _:
        print("Invalid day")
````

````python
day = 7
match day:
    case 2 | 3 | 4 | 5 | 6:
        print("Weekday")
    case 1 | 7:
        print("Weekend")
    case _:
        print("Invalid day")
````
