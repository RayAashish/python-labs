# Credit Card Validator
**Given  sting of digits for a credit card
number,determie if it's a valid card number
using the following method:**
1. Starting from the second-to-last digit,
double every other digit moving left.
3. If doubling a digit results in a number
greater than 9, subtract 9.
4. Sum all the digits (doubled and undoubled).
5. If the total is divisible by 10,the number
is valid.

```python
# card = "4235 5900 1093 3819"
card = "4162 1100 7113 1760"
card = card.replace(" ", '')
total = 0
l = len(card)
for i in range (l):
    num = int(card[i])
    if (i % 2 == 0):
        num *= 2
        if (num > 9):
            num -= 9
    total += num;

print(total)
if (total % 10 == 0):
    print("It's a valid credit card")
else:
    print("It's not a valid credit card")
```
