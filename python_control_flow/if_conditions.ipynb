# Control Flow
**Control flow decides which code executes, when it executes & how many times it executes<br>**
**Things covered<br>**
1. if condition
2. if elif else condition
3. nested if condition

```python
# if else is executed when a code depends upon a condition
age = int(input("Enter your age: "))
if (age >= 18):
    print("You're eligible for vote")
else:
    print("You're not eligible for vote")
```

```python
# if elif else condition
mark = int(input("Enter your marks : "))
if (mark >= 90):
    print("You've got O grade")
elif (mark >= 80):
    print("You've got A grade")
elif (mark >= 70):
    print("You've got B grade")
elif (mark >= 60):
    print("You've got C grade")
elif (mark >= 50):
    print("You've got D grade")
elif (mark >= 40):
    print("You've got P grade")
else:
    print("You've failed")
```

```python
# nested if condition
username = 'admin'
password = '12345'
if (username == 'admin'):
    if (password == '12345'):
        print("Login successfull")
    else:
        print("Wrong password")
else:
    print("Invalid user")
```
